## Summary
We consider for a matrix $A \in \mathbb{R}^{m\times n}$ and $d\in \mathbb{R}^n$ the problem of determining whether the system of linear equations $Ax=d$ has a unique integer solution $x\in \{-1,1\}^n$. A natural relaxation of the problem is to consider the linear program (LP)
\begin{align}
    \min & \quad \|x\|_\infty \\
    \text{s.t.} & \quad Ax=d \,, \\
    & \quad x \in [-1,1]^n \,.
\end{align}
In $[1]$ necessary and sufficient conditions are derived for which the LP recovers a unique integer solution. Furthermore using an argument by Wendel $[2]$ (a probabilistic version of Cover's Theorem) it is shown that for a generic random matrix $A$ (independent columns, that are symmetrically distributed) the probability of recovering a unique integer solution is \textit{exactly} $2^{-n+1} \sum^{m-1}_{i=0}\binom{n-1}{i}$. Defining $\gamma:= m/n$ and and given an integer $x_0$ such that $Ax_0=d$ one can use Hoeffding's inequality to show that asymptotically for $\gamma>1/2$ the probability of recovering the integer solution $x_0$ via the LP tends to $1$ and for $\gamma<1/2$ the probability tends to $0$. Based on an idea by Pravesh Kothari we consider the semidefinite programming (SDP) relaxation
\begin{align}
    \min_X & \quad \|X\|_1 \\
    \text{s.t.} & \quad AXA^{\top}=dd^{\top} \,, \\
    & \quad X_{ii} = 1 \quad \text{ for } i=1, \ldots, n \,, \\
    &\quad X\succeq 0\,,
\end{align}
and empirically check whether we can increase the range of $\gamma$ (meaning a threshold smaller than $1/2$) for which we can recover the integer solution $x_0$ to the problem $Ax= d$. More precisely we check whether $x_0$ is an eigenvector corresponding to the leading eigenvalue of the returned matrix $X$. Empirically we find that the SDP seems to also come with the critical threshold of $1/2$.
The main question to be answered is the following: Is it possible to decrease the threshold of $1/2$ by means of a polynomial time algorithm?
I also list some questions that arose, but that I could not answer:
1. When the LP recovers the unique integer solution, does the SDP do so as well? Empirically this seems to be the case and also practically one can enforce the LP to be contained in the SDP, however I did not prove it.
2. Can one prove that the SDP relaxation comes with the same threshold $1/2$? For the case of a Gaussian matrix $A$ I proved that the threshold is at least $1/\sqrt{2}$.
