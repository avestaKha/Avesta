# Problem 1.1

For $n \in \mathbb{N}$ define $B^{(1)}_{n}=A_n$ and $B^{(2)}_{n}=A_n^{c}$.
Then for each $k \in \mathbb{N}$ the family of $2^k$ events
$$
\left[
E_{i_1,\dots,i_k}\;:=\;\bigcap_{n=1}^{k} B^{(i_n)}_{n},
\qquad i_n \in \{1,2\}
\right]
$$
forms a disjoint partition of $\Omega$ and, by independence, each event has probability $2^{-k}$.
Now any $\omega \in \Omega$ is, for each $k \in \mathbb{N}$, contained in such an event and thereby
$$
P(\{\omega\}) \le 2^{-k}\ \text{for all } k \in \mathbb{N}
\;\Rightarrow\;
P(\{\omega\})=0,
$$
so no point masses can exist.
